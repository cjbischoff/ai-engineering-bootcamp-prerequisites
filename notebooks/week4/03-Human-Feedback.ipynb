# Import Dependencies 

```python
from langsmith import Client
```

# Attach Feedback to a specific trace in Langsmith

019c7305-ec50-76b2-bb97-d3ae5d0fba6a

```python
client = Client()
```

```python
RUN_ID = "019c7305-ec50-76b2-bb97-d3ae5d0fba6a"
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_source_type="api",
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comments",
    value="This a long form free text feedback left by Christopher",
    feedback_source_type="api",
)
```

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comments_with_score",
    value="This a long form free text feedback left by Christopher",
    score=0,
    feedback_source_type="api",
)
```
